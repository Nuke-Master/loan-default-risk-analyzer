```python
import pandas as pd
import sqlite3

df = pd.read_csv("../data/loan_data.csv")

df.head()
df.shape
connection = sqlite3.connect("../data/loan_risk.db")
df.to_sql("loans", connection, if_exists="replace", index=False)
query = """
SELECT *
FROM loans
LIMIT 5;
"""

result = pd.read_sql_query(query, connection)

result

query = """
SELECT COUNT(*) AS total_loans
FROM loans;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    "not.fully.paid",
    COUNT(*) AS loan_count
FROM loans
GROUP BY "not.fully.paid";
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    AVG("not.fully.paid") * 100 AS non_repayment_rate
FROM loans;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,
    COUNT(*) AS total_loans
FROM loans
GROUP BY purpose
ORDER BY total_loans DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,
    ROUND(AVG("int.rate") * 100, 2) AS avg_interest_rate
FROM loans
GROUP BY purpose
ORDER BY avg_interest_rate DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,
    COUNT(*) AS total_loans,
    SUM("not.fully.paid") AS unpaid_loans,
    ROUND(AVG("not.fully.paid") * 100, 2) AS non_repayment_rate
FROM loans
GROUP BY purpose
ORDER BY non_repayment_rate DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    "not.fully.paid",
    COUNT(*) AS total_loans,
    ROUND(AVG(fico), 2) AS average_fico
FROM loans
GROUP BY "not.fully.paid";
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    "not.fully.paid",
    COUNT(*) AS total_loans,
    ROUND(AVG(dti), 2) AS average_dti
FROM loans
GROUP BY "not.fully.paid";
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    "not.fully.paid",
    COUNT(*) AS total_loans,
    ROUND(AVG("int.rate") * 100, 2) AS average_interest_rate
FROM loans
GROUP BY "not.fully.paid";
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    "credit.policy",
    COUNT(*) AS total_loans,
    ROUND(AVG("not.fully.paid") * 100, 2) AS non_repayment_rate
FROM loans
GROUP BY "credit.policy"
ORDER BY "credit.policy";
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,
    COUNT(*) AS total_loans,
    SUM("not.fully.paid") AS unpaid_loans,
    ROUND(AVG("not.fully.paid") * 100, 2) AS non_repayment_rate
FROM loans
GROUP BY purpose
ORDER BY non_repayment_rate DESC;
"""

result = pd.read_sql_query(query, connection)

result

connection = sqlite3.connect("../data/loan_risk.db")

df.to_sql("loans", connection, if_exists="replace", index=False)

query = """
SELECT *
FROM loans
LIMIT 5;
"""

pd.read_sql_query(query, connection)

query = """
SELECT COUNT(*) AS total_loans
FROM loans;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    "not.fully.paid",
    COUNT(*) AS loan_count
FROM loans
GROUP BY "not.fully.paid";
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    ROUND(AVG("not.fully.paid") * 100, 2) AS non_repayment_rate
FROM loans;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,
    COUNT(*) AS total_loans
FROM loans
GROUP BY purpose
ORDER BY total_loans DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,
    COUNT(*) AS total_loans,
    SUM("not.fully.paid") AS unpaid_loans,
    ROUND(AVG("not.fully.paid") * 100, 2) AS non_repayment_rate
FROM loans
GROUP BY purpose
ORDER BY non_repayment_rate DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    "not.fully.paid",
    COUNT(*) AS total_loans,
    ROUND(AVG(fico), 2) AS average_fico
FROM loans
GROUP BY "not.fully.paid";
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    CASE
        WHEN fico < 650 THEN 'Below 650'
        WHEN fico BETWEEN 650 AND 699 THEN '650-699'
        WHEN fico BETWEEN 700 AND 749 THEN '700-749'
        ELSE '750+'
    END AS fico_group,
    
    COUNT(*) AS total_loans,
    
    SUM("not.fully.paid") AS unpaid_loans,
    
    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate

FROM loans

GROUP BY fico_group

ORDER BY non_repayment_rate DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    CASE
        WHEN dti < 10 THEN 'Low DTI'
        WHEN dti BETWEEN 10 AND 20 THEN 'Medium DTI'
        WHEN dti BETWEEN 20 AND 30 THEN 'High DTI'
        ELSE 'Very High DTI'
    END AS dti_group,

    COUNT(*) AS total_loans,

    SUM("not.fully.paid") AS unpaid_loans,

    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate

FROM loans

GROUP BY dti_group

ORDER BY non_repayment_rate DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    COUNT(*) AS high_risk_loans,
    SUM("not.fully.paid") AS unpaid_loans,
    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate

FROM loans

WHERE fico < 650
AND dti > 20;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    COUNT(*) AS high_interest_loans,
    SUM("not.fully.paid") AS unpaid_loans,
    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate

FROM loans

WHERE "int.rate" > 0.15;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,
    COUNT(*) AS total_loans,
    SUM("not.fully.paid") AS unpaid_loans,
    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate,
    ROUND(AVG(fico), 2) AS average_fico,
    ROUND(AVG(dti), 2) AS average_dti

FROM loans

WHERE purpose = 'small_business'

GROUP BY purpose;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,

    COUNT(*) AS total_loans,

    SUM(
        CASE
            WHEN "not.fully.paid" = 0 THEN 1
            ELSE 0
        END
    ) AS fully_paid,

    SUM(
        CASE
            WHEN "not.fully.paid" = 1 THEN 1
            ELSE 0
        END
    ) AS not_fully_paid,

    ROUND(
        AVG("int.rate") * 100,
        2
    ) AS avg_interest_rate,

    ROUND(
        AVG(fico),
        2
    ) AS avg_fico,

    ROUND(
        AVG(dti),
        2
    ) AS avg_dti

FROM loans

GROUP BY purpose

ORDER BY total_loans DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    purpose,
    COUNT(*) AS total_loans,
    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate

FROM loans

GROUP BY purpose

HAVING COUNT(*) >= 100

ORDER BY non_repayment_rate DESC

LIMIT 5;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    CASE
        WHEN "inq.last.6mths" = 0 THEN 'No inquiries'
        WHEN "inq.last.6mths" BETWEEN 1 AND 2 THEN '1-2 inquiries'
        WHEN "inq.last.6mths" BETWEEN 3 AND 5 THEN '3-5 inquiries'
        ELSE '6+ inquiries'
    END AS inquiry_group,

    COUNT(*) AS total_loans,

    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate

FROM loans

GROUP BY inquiry_group

ORDER BY non_repayment_rate DESC;
"""

pd.read_sql_query(query, connection)


query = """
SELECT
    CASE
        WHEN fico < 650 AND dti > 20 THEN 'High Risk'
        WHEN fico < 700 OR dti > 20 THEN 'Moderate Risk'
        ELSE 'Lower Risk'
    END AS risk_group,

    COUNT(*) AS total_loans,

    SUM("not.fully.paid") AS unpaid_loans,

    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate,

    ROUND(
        AVG(fico),
        2
    ) AS average_fico,

    ROUND(
        AVG(dti),
        2
    ) AS average_dti,

    ROUND(
        AVG("int.rate") * 100,
        2
    ) AS average_interest_rate

FROM loans

GROUP BY risk_group

ORDER BY non_repayment_rate DESC;
"""

pd.read_sql_query(query, connection)

query = """
SELECT
    CASE
        WHEN "pub.rec" = 0 THEN 'No public records'
        ELSE 'Has public records'
    END AS public_record_status,

    COUNT(*) AS total_loans,

    ROUND(
        AVG("not.fully.paid") * 100,
        2
    ) AS non_repayment_rate

FROM loans

GROUP BY public_record_status

ORDER BY non_repayment_rate DESC;
"""

pd.read_sql_query(query, connection)
```
